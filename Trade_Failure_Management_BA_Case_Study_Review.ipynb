## 2. Complete Business Analysis Case Study

# Trade Failure Management System — Business Analysis Case Study

## 1. Case-study context

This project is a **simple learning and portfolio case study** designed to practise Business & Technical Analysis in a fictional investment-banking technology environment.

The scenario and data are synthetic. No confidential banking information is used, and the project does not claim to reproduce the internal processes or policies of any real bank.

---

## 2. Business problem

A fictional Trade Operations team investigates failed trade settlements. The current investigation process is fragmented across several internal systems, files and spreadsheets.

This makes it difficult to maintain consistent information, determine who owns each investigation, prioritise urgent cases and provide management with a single view of outstanding failures.

### Business objective

Create a clearer case-management approach that improves:

- accountability;
- investigation traceability;
- prioritisation;
- status visibility; and
- management oversight.

---

## 3. Stakeholders

### Trade Operations Analysts
Primary users who investigate failed trades, record findings, update case status and work towards resolution.

### Operations Manager
Needs visibility of outstanding failures, workload, priorities, ownership and progress.

### Front Office / Trading Teams
May provide trade information or clarification required during an investigation.

### Technology / Development Team
Would translate approved requirements into technical functionality and support the resulting system.

### Risk & Compliance
Interested in appropriate auditability, traceability and controlled handling of operational records.

---

## 4. AS-IS process

The fictional current process is:

```text
Failed-trade report received
        ↓
Analyst identifies a failed trade
        ↓
Checks several internal systems
        ↓
Checks supporting spreadsheets/files
        ↓
Investigates the failure
        ↓
Records investigation notes manually
        ↓
Attempts resolution
```

The purpose of the AS-IS review is to understand the current workflow **before** proposing technology.

---

## 5. Pain points

### P1 — Fragmented information
Trade information and investigation records are spread across several systems and spreadsheets, increasing the risk of inconsistent or mismatched information.

### P2 — Manual investigation records
Investigation notes are recorded manually, which can reduce consistency and make change history harder to trace.

### P3 — Unclear ownership
There is no sufficiently clear ownership mechanism. More than one analyst could unknowingly investigate the same failed trade, duplicating effort.

### P4 — Inconsistent prioritisation
High-value, older or otherwise important failed trades may not be prioritised consistently.

### P5 — Limited management visibility
Management lacks a single consolidated view of outstanding failed trades, their status, priority and owner.

---

## 6. Requirements

The pain points are translated into functional needs before selecting specific products or technologies.

| ID | Functional requirement | Related pain point |
|---|---|---|
| FR1 | The system shall maintain a central record of identified failed trades and relevant trade details. | P1 |
| FR2 | Analysts shall be able to record, update and review investigation notes with a history of changes. | P2 |
| FR3 | Each failed-trade case shall have one clearly assigned current owner. | P3 |
| FR4 | The system shall provide a central view of outstanding cases including status, owner, age and priority. | P5 |
| FR5 | Failed trades shall be prioritised using agreed criteria such as trade value, age and business impact. | P4 |
| FR6 | Analysts shall be able to update a case through its investigation lifecycle and record the final resolution/root cause. | P2, P5 |

### Requirement vs solution

For example:

**Problem:** Management cannot easily see all outstanding failed trades.  
**Requirement:** Management must be able to monitor outstanding cases centrally.  
**Possible solution:** A dashboard or central case-management view.

This distinction avoids choosing technology before the business need is understood.

---

## 7. TO-BE process

```text
Failed trade identified
        ↓
Central case created
        ↓
Priority assigned using agreed rules
        ↓
One analyst assigned as current owner
        ↓
Investigation performed
        ↓
Investigation notes recorded against the case
        ↓
Status updated
        ↓
Trade resolved
        ↓
Resolution / root cause recorded
        ↓
Management view updated
```

The proposed process addresses the main AS-IS issues by centralising case information and making ownership, priority and progress visible.

---

## 8. User stories

### US-01 — Clear ownership

**As an Operations Manager, I want each failed-trade case to have one clearly assigned owner, so that accountability is clear and duplicate investigations are avoided.**

### US-02 — Central visibility

**As an Operations Manager, I want to see all outstanding failed trades in one central view, so that I can monitor progress and identify urgent cases requiring attention.**

### US-03 — Investigation status

**As a Trade Operations Analyst, I want to update the status of a failed-trade case, so that other users can understand where the investigation currently stands.**

### US-04 — Investigation notes

**As a Trade Operations Analyst, I want to record investigation notes against a case, so that findings and actions remain traceable throughout the investigation.**

---

## 9. Acceptance criteria

### US-01 — Ownership

**AC1**  
Given a new failed-trade case has been created,  
When the case is assigned to an analyst,  
Then that analyst shall be displayed as the current owner.

**AC2**  
Given a case already has an owner,  
When another analyst views the case,  
Then the current owner shall be visible.

**AC3**  
Given ownership needs to change,  
When the case is reassigned,  
Then the new analyst shall become the current owner and the previous ownership shall remain available in history.

### US-02 — Central visibility

**AC1**  
Given outstanding failed-trade cases exist,  
When an Operations Manager opens the central view,  
Then the outstanding cases shall be displayed.

**AC2**  
Each displayed case shall show at least its status, current owner and priority.

---

## 10. MoSCoW prioritisation

| Capability | Priority | Rationale |
|---|---|---|
| Central failed-trade record | **Must** | Foundation of consistent case management |
| One current owner per case | **Must** | Establishes accountability and reduces duplicate investigation |
| Investigation status tracking | **Must** | Makes progress visible |
| Prioritisation using agreed criteria | **Must** | Helps urgent cases receive appropriate attention |
| Investigation notes/history | **Must** | Supports traceability |
| Management dashboard/central view | **Should** | Important for oversight, but the core workflow can be established first |
| AI/ML-assisted validation | **Could / Future** | Consider only after the core process and reliable data exist |

---

## 11. UAT

User Acceptance Testing checks whether the proposed functionality meets the fictional Operations team's needs. These tests are **designed for the case study; they have not been executed against a production banking system**.

### UAT-01 — Ownership and reassignment

**Steps**
1. Create a failed-trade case.
2. Assign it to Analyst A.
3. Verify Analyst A appears as current owner.
4. Reassign the case to Analyst B.
5. Verify Analyst B becomes current owner.
6. Inspect ownership history.

**Expected result:** Analyst B is the only current owner and Analyst A remains visible in ownership history.

### UAT-02 — Illustrative high-value prioritisation

For demonstration only, assume trades above **£1 million** are classified as High Priority.

Create:
- Trade A — £10,000
- Trade B — £5,000,000
- Trade C — £50,000

**Expected result:** Trade B is High Priority; Trades A and C are not High under this illustrative rule.

The £1 million threshold is **not a real bank policy**. In a real project, prioritisation rules would be elicited, agreed and approved by stakeholders.

More detail is available in `docs/uat_test_cases.md`.

A priority field is intentionally absent from the raw dataset because prioritisation should follow an agreed business rule rather than being assumed to be part of the source data.

## 14. Future AI/ML enhancement

An early idea in the case study was to use AI/ML to help validate investigation records or identify recurring failure patterns.

That idea is deliberately **not included in the first-release requirements**.

A more appropriate sequence is:

```text
Establish reliable central workflow
        ↓
Collect consistent historical case data
        ↓
Measure data quality and operational patterns
        ↓
Identify a specific automation/ML use case
        ↓
Evaluate model benefit, risk and controls
```

Potential future examples could include:

- flagging inconsistent investigation records;
- identifying recurring failure patterns;
- suggesting similar historical cases; or
- supporting risk-based prioritisation.

Any such capability would require separate requirements, validation, governance and human oversight.

## 15. What I learned from the case study

This exercise demonstrates the difference between immediately proposing a technical solution and first understanding a business problem.

The main BA flow used was:

**Business Problem → Stakeholders → AS-IS → Pain Points → Requirements → TO-BE → User Stories → Acceptance Criteria → MoSCoW → UAT → Backlog → Future Enhancement**

It also demonstrates why ownership, traceability and prioritisation should be defined as business requirements before deciding whether the implementation should use a specific ticketing platform, dashboard technology or AI/ML model.
